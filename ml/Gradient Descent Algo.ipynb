```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
# Define function
def f(x):
    return (x + 3)**2

# Define derivative
def df(x):
    return 2 * (x + 3)
```

```python
# Initial values
x = 2
learning_rate = 0.1
iterations = 100

# Gradient Descent
for i in range(iterations):

    gradient = df(x)

    x = x - learning_rate * gradient

    print("Iteration:", i+1, "x =", x, "y =", f(x))

print("\nLocal Minimum:")
print("x =", x)
print("y =", f(x))
```

```python
# Generate x values
x_values = np.linspace(-6, 4, 100)

# Calculate y values
y_values = f(x_values)

# Plot function
plt.plot(x_values, y_values, label='y = (x+3)^2')

# Mark minimum point
plt.scatter(-3, 0, color='red', label='Minimum Point')

plt.xlabel("x")
plt.ylabel("y")
plt.title("Gradient Descent - Local Minimum")

plt.legend()
plt.grid()
plt.show()
```

```python
x = 2
learning_rate = 0.1

x_history = []

for i in range(20):

    x_history.append(x)

    x = x - learning_rate * df(x)

plt.plot(x_history, marker='o')

plt.xlabel("Iterations")
plt.ylabel("Value of x")
plt.title("Gradient Descent Convergence")

plt.grid()
plt.show()
```
